```python
## data preprocessing , cleaning, null-handling delete for "ids_allcountries_data" FILE

#STEP 1 Formatting column names 
ids_allcountries_data.columns = (
    ids_allcountries_data.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
    .str.replace("-", "_", regex=False)
)

print(ids_allcountries_data.columns.tolist())
```

```python
# Step 2 — Remove 2027–2032

future_years = [str(year) for year in range(2027, 2033)]

ids_allcountries_data = ids_allcountries_data.drop(
    columns=[year for year in future_years if year in ids_allcountries_data.columns]
)
```

```python
#Step 3 — Convert data column-wise to row-wise,

id_columns = [
    "country_name",
    "country_code",
    "counterpart_area_name",
    "counterpart_area_code",
    "series_name",
    "series_code"
]

year_columns = [
    str(year)
    for year in range(2000, 2027)
    if str(year) in ids_allcountries_data.columns
]

df_long = ids_allcountries_data.melt(
    id_vars=id_columns,
    value_vars=year_columns,
    var_name="year",
    value_name="debt_value"
)
```

```python
df_long.head(10)
```

```python
## changed column data type for (year , debt_value)
df_long["year"] = pd.to_numeric(
    df_long["year"],
    errors="coerce"
)

df_long["debt_value"] = pd.to_numeric(
    df_long["debt_value"],
    errors="coerce"
)
```

```python
df_long.dtypes
```

```python
## NULL VALUE CHECKING

df_long.isnull().sum()
```

```python
df_long = df_long.sort_values(
    ["country_code", "series_code", "year"]
).reset_index(drop=True)
```

```python
## APPLY ffill() + bfill()

df_long["debt_value"] = (
    df_long
    .groupby(["country_code", "series_code"])["debt_value"]
    .transform(lambda x: x.ffill().bfill())
)
```

```python
df_long[
    df_long[
        [
            "country_name",
            "country_code",
            "series_name",
            "series_code",
            "debt_value"
        ]
    ].isnull().all(axis=1)
].shape
```

```python
df_long[
    df_long["country_code"].isnull()
][
    [
        "country_name",
        "country_code",
        "counterpart_area_name",
        "counterpart_area_code",
        "series_name",
        "series_code",
        "year",
        "debt_value"
    ]
].head(20)
```

```python
## Remove all rows where country_code is NULL 

df_long = df_long.dropna(
    subset=["country_code"]
).reset_index(drop=True)
```

```python
# Remove completely empty groups
df_long = df_long[
    df_long.groupby(
        ["country_code", "counterpart_area_code", "series_code"]
    )["debt_value"]
    .transform("count") > 0
].reset_index(drop=True)
```

```python
q26 = pd.read_sql("""
    WITH country_debt AS
    (
        SELECT
            country_name,
            SUM(debt_value) AS total_debt
        FROM allcountries_debt_data
        WHERE country_name IS NOT NULL
          AND debt_value IS NOT NULL
        GROUP BY country_name
    ),
    debt_stats AS
    (
        SELECT
            AVG(total_debt) AS avg_debt,
            STDDEV(total_debt) AS std_debt
        FROM country_debt
    )
    SELECT
        c.country_name,
        c.total_debt,
        CASE
            WHEN c.total_debt > s.avg_debt + s.std_debt
                THEN 'High Debt'
            WHEN c.total_debt >= s.avg_debt
                THEN 'Medium Debt'
            ELSE 'Low Debt'
        END AS debt_category
    FROM country_debt c
    CROSS JOIN debt_stats s
    ORDER BY c.total_debt DESC;
""", engine)

display(q26)
```
